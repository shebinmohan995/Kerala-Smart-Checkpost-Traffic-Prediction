```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_train.shape
```

```python
x.drop('Vehicle_ID',axis=1)
```

```python
x_train=x_train.drop('Vehicle_ID',axis=1)
```

```python
x_test=x_test.drop('Vehicle_ID',axis=1)
```

```python
from sklearn.preprocessing import OneHotEncoder
coder=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
```

```python
cat=[col for col in cat if col !='Vehicle_ID']
```

```python
x_train_cat=coder.fit_transform(x_train[cat])
x_test_cat=coder.transform(x_test[cat])
```

```python
num=x_train.select_dtypes(include=['int64','float64']).columns.tolist()
num
```

```python
x_test_num=x_test[num].to_numpy()
x_train_num=x_train[num].to_numpy()
```

```python
x_train_processed=np.concatenate([x_train_num,x_train_cat],axis=1)
```

```python
x_test_processed=np.concatenate([x_test_num,x_test_cat],axis=1)
```

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier(
    max_depth=10,
    random_state=42)
model.fit(x_train_processed,y_train)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
y_pred=model.predict(x_test_processed)
ac=accuracy_score(y_test,y_pred)
ac
```

```python
cr=classification_report(y_test,y_pred)
cr
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
plt.Figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
plt.xlabel('Predicted Traffic')
plt.ylabel('Actual Traffic')
plt.title('Traffic Density Prediction')
```

```python
from sklearn.tree import plot_tree
plt.Figure(figsize=(50,15))
plot_tree(model,filled=True,rounded=True,max_depth=3,fontsize=6,precision=1)
plt.tight_layout(pad=0.2)
```
